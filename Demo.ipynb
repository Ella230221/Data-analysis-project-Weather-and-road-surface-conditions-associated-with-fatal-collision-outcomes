# Canadian road safety in 2019
Open from the repository root. The stored results are sufficient to walk through the analysis. The final cell reruns Spark and the models when the official CSV and dependencies are installed.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

results = Path('results')
display(json.loads((results / 'audit.json').read_text()))
display(pd.read_csv(results / 'weather_surface.csv').head(12))
```

## Environmental associations
Each rate is the fatal count divided by the number of included collisions in the category. The horizontal intervals show Wilson uncertainty.

```python
display(pd.read_csv(results / 'environment.csv'))
display(Image(filename=str(results / 'environment.png')))
```

## Held-out logistic comparison
M0 uses training prevalence, M1 uses controls, and M2 adds weather and surface to the same controls.

```python
display(pd.read_csv(results / 'metrics.csv').round(4))
display(pd.read_csv(results / 'metrics_known.csv').round(4))
display(Image(filename=str(results / 'precision_recall.png')))
```

## Spark partition experiment
Every measured run computes the same complete grouped result.

```python
display(pd.read_csv(results / 'timing_runs.csv').round(3))
display(pd.read_csv(results / 'timing_summary.csv').round(3))
display(Image(filename=str(results / 'spark_timing.png')))
```
